```python
def factorial(n):
  if (n==0) or (n==1):
    return 1
  else:
    return (n*factorial(n-1))
num=5
print("the number is:", num)
print("the factorial is:", factorial(num))
```

```python
import pandas as pd
df= pd.DataFrame({"car":["bmw","volvo"],"passings":[5,6]})
print(df)
print("\n")
df1= pd.DataFrame({"car":["honda","hyundai"],"passings":[3,4]})
print(df1)
print("\n")
df2= df._append(df1,ignore_index=True)  ## in append function we can add only two tables
print(df2)
print("\n")
df3=pd.concat([df,df1,df2],ignore_index=True)  ## in concat function we can add multiple tables
print(df3)
```
